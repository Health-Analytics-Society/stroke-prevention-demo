```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("data/raw/stroke_data.csv")
df.head()

```

## 1) Target distribution
We start here to understand class imbalance and why accuracy can be misleading later.


```python
# Counts and proportions of the target
counts = df["stroke"].value_counts()
pct = df["stroke"].value_counts(normalize=True).round(4)
counts, pct
```

## 2) Missingness
Missingness can carry signal in health data. Identify columns to drop/impute/flag.


```python
missing_pct = (df.isna().mean() * 100).sort_values(ascending=False).round(2)
missing_pct.head(20)
```

## 3) Feature vs target (quick signal check)
We look for features that differ between stroke vs non-stroke groups.


```python
df.groupby("stroke")["age"].mean() if "age" in df.columns else "No age column"
```

## Takeaways
- Target is imbalanced
- Some features show separation
- Missingness likely needs a consistent strategy
