# Data Sanity Check (DP-3)

Confirms the dataset loads correctly and prints basic statistics before trusting any model results.

```python
import pandas as pd
from pathlib import Path

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent

DATA_PATH = ROOT / "data" / "raw" / "stroke_data.csv"
if not DATA_PATH.exists():
    raise FileNotFoundError(f"Could not find dataset at {DATA_PATH}")

df = pd.read_csv(DATA_PATH)
print(f"Dataset loaded from: {DATA_PATH}")
```

```python
# Number of rows and columns
print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")
```

```python
# Stroke label counts and percentage
print("Stroke label counts:")
print(df["stroke"].value_counts(dropna=False))
print(f"\nPercent positive (stroke=1): {df['stroke'].mean() * 100:.2f}%")
```

```python
# Full list of column names
print("Column names:")
for col in df.columns:
    print(f"  {repr(col)}")
```

```python
# Min and max for 5 numeric columns
num_cols = df.select_dtypes(include="number").columns.tolist()
selected = num_cols[:5]
print("Min and max for 5 numeric columns:")
for col in selected:
    print(f"  {col}: min={df[col].min()}, max={df[col].max()}")
```
